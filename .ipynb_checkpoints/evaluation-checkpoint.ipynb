## Evaluation Setup

On recharge `typeR` depuis le CSV original car il n'a pas été gardé dans le preprocessing.

```python
# Recharger typeR depuis le CSV original
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

meta = pd.read_csv('data/Tripadvisor.csv', low_memory=False)
meta = meta[['id', 'typeR', 'activiteSubCategorie', 'restaurantTypeCuisine']]

# Convertir les colonnes liste (même logique que ta camarade)
for col in ['activiteSubCategorie', 'restaurantTypeCuisine']:
    meta[col] = meta[col].apply(
        lambda s: [int(x.strip()) for x in str(s).split(',') if x.strip() not in ['nan', '']]
    )

print(meta.shape)
print(meta['typeR'].value_counts())
meta.head()
```

## Train / Test Split 50/50

On split les **places** (pas les reviews). Le train = requêtes, le test = corpus dans lequel on cherche.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

train_df, test_df = train_test_split(reviews_for_tfidf, test_size=0.5, random_state=42)
train_df = train_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

print(f"Train : {len(train_df)} places")
print(f"Test  : {len(test_df)} places")
```

```python
# TF-IDF : fit sur train, transform sur test
tfidf_eval = TfidfVectorizer()
train_matrix = tfidf_eval.fit_transform(train_df['text_resume'])
test_matrix  = tfidf_eval.transform(test_df['text_resume'])

# Matrice similarité : shape (nb_train x nb_test)
sim_matrix = cosine_similarity(train_matrix, test_matrix)
print("Matrice similarité shape :", sim_matrix.shape)
```

## Ranking Error — Level 1 (typeR)

Pour chaque place du train, on classe tout le test par similarité décroissante.  
On note la **position** de la première place du test qui a le même `typeR`.  
- Position 0 = recommandation parfaite (erreur = 0)  
- Si aucune place du test n'a le même type → requête ignorée

```python
# Joindre les metadata
train_meta = train_df.merge(meta[['id','typeR']], left_on='idplace', right_on='id', how='left')
test_meta  = test_df.merge(meta[['id','typeR']], left_on='idplace', right_on='id', how='left')

errors_l1 = []

for i in range(len(train_df)):
    query_type = train_meta.iloc[i]['typeR']

    # Ignorer si pas de typeR ou aucun match possible dans le test
    if pd.isna(query_type):
        continue
    if query_type not in test_meta['typeR'].values:
        continue

    # Classer le test par similarité décroissante
    ranked = sim_matrix[i].argsort()[::-1]

    for rank, idx in enumerate(ranked):
        if test_meta.iloc[idx]['typeR'] == query_type:
            errors_l1.append(rank)
            break

print(f"Nb requêtes évaluées : {len(errors_l1)}")
print(f"Mean Ranking Error Level 1 : {sum(errors_l1)/len(errors_l1):.2f}")
print(f"Erreur médiane : {sorted(errors_l1)[len(errors_l1)//2]}")
```

## Ranking Error — Level 2 (sous-catégories)

Évaluation plus fine :  
- Attractions → `activiteSubCategorie`  
- Restaurants → `restaurantTypeCuisine`  

**Match positif** si au moins une catégorie est commune entre la requête et la place recommandée.

```python
train_meta2 = train_df.merge(meta, left_on='idplace', right_on='id', how='left')
test_meta2  = test_df.merge(meta, left_on='idplace', right_on='id', how='left')


def get_cats_l2(row):
    """Retourne les catégories fines selon le type de lieu."""
    t = row.get('typeR', None)
    if t in ('A', 'AP'):
        c = row.get('activiteSubCategorie', [])
    elif t == 'R':
        c = row.get('restaurantTypeCuisine', [])
    else:
        return []  # Hotels : pas de sous-catégorie fine disponible ici
    return c if isinstance(c, list) else []


errors_l2 = []

for i in range(len(train_df)):
    query_cats = get_cats_l2(train_meta2.iloc[i])

    if not query_cats:
        continue

    # Vérifier qu'un match existe dans le test
    test_cats_list = [get_cats_l2(test_meta2.iloc[j]) for j in range(len(test_df))]
    query_set = set(query_cats)
    if not any(query_set & set(c) for c in test_cats_list):
        continue

    ranked = sim_matrix[i].argsort()[::-1]

    for rank, idx in enumerate(ranked):
        if query_set & set(test_cats_list[idx]):
            errors_l2.append(rank)
            break

print(f"Nb requêtes évaluées : {len(errors_l2)}")
print(f"Mean Ranking Error Level 2 : {sum(errors_l2)/len(errors_l2):.2f}")
print(f"Erreur médiane : {sorted(errors_l2)[len(errors_l2)//2]}")
```

## Visualisation des résultats

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Distribution des erreurs Level 1
axes[0].hist(errors_l1, bins=30, color='steelblue', edgecolor='white')
axes[0].axvline(sum(errors_l1)/len(errors_l1), color='red', linestyle='--',
                label=f'Mean = {sum(errors_l1)/len(errors_l1):.1f}')
axes[0].set_title('Distribution Ranking Error — Level 1 (typeR)')
axes[0].set_xlabel('Position du 1er match')
axes[0].set_ylabel('Nb requêtes')
axes[0].legend()

# Distribution des erreurs Level 2
axes[1].hist(errors_l2, bins=30, color='coral', edgecolor='white')
axes[1].axvline(sum(errors_l2)/len(errors_l2), color='red', linestyle='--',
                label=f'Mean = {sum(errors_l2)/len(errors_l2):.1f}')
axes[1].set_title('Distribution Ranking Error — Level 2 (sous-catégories)')
axes[1].set_xlabel('Position du 1er match')
axes[1].set_ylabel('Nb requêtes')
axes[1].legend()

plt.tight_layout()
plt.savefig('ranking_error_distribution.png', dpi=150)
plt.show()
```

## Interprétation

**Level 1** évalue si le modèle retrouve un lieu du même type général (hôtel, restaurant, attraction). C'est le critère le plus simple — une erreur faible ici confirme que le vocabulaire utilisé dans les reviews est suffisamment distinctif selon le type de lieu.

**Level 2** est plus exigeant : il vérifie si la sous-catégorie précise est retrouvée. L'erreur est naturellement plus haute car le critère est plus strict. Un match est considéré positif si au moins une sous-catégorie est commune entre la requête et la recommandation.

Ces résultats permettent de valider (ou non) l'hypothèse centrale du projet : **des expériences similaires sont décrites avec des mots similaires**.
