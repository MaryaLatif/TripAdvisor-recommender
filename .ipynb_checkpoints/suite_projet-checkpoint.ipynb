## Train / Test Split

On divise les **places** (pas les reviews) en deux moitiés : 50% train (requêtes) et 50% test (corpus de recommandation).  
Le système cherche dans le test set les places les plus similaires à chaque place du train set.

**Attention** : les métadonnées (`typeR`, `activiteSubCategorie`, etc.) servent **uniquement** à l'évaluation, jamais en entrée du modèle.

````python
from sklearn.model_selection import train_test_split

# On part du dataframe reviews_for_tfidf qui a une ligne par place
train_df, test_df = train_test_split(reviews_for_tfidf, test_size=0.5, random_state=42)

train_df = train_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

print(f"Train : {len(train_df)} places")
print(f"Test  : {len(test_df)} places")
````

## Ranking Error — définition

Pour chaque place du **train**, on classe toutes les places du **test** par similarité décroissante.  
On cherche la **position** de la première place qui partage la même catégorie que la requête.

- Position 0 → recommandation parfaite (erreur = 0)
- Position k → erreur = k
- Aucune place dans le test avec la même catégorie → la requête est ignorée

On calcule ensuite la **moyenne** de ces erreurs (Mean Ranking Error).

## TF-IDF sur train/test séparés

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# On fit le vectorizer sur le train uniquement
tfidf_vec = TfidfVectorizer()
train_matrix = tfidf_vec.fit_transform(train_df["text_resume"])
test_matrix  = tfidf_vec.transform(test_df["text_resume"])

# Matrice de similarité : train (requêtes) × test (corpus)
sim_train_test = cosine_similarity(train_matrix, test_matrix)

print("Shape similarité train×test :", sim_train_test.shape)
````

## Ranking Error Level 1 — basé sur `typeR`

`typeR` peut valoir : **H** (Hotel), **R** (Restaurant), **A** (Attraction), **AP** (Attraction Product).

````python
def ranking_error_level1(train_df, test_df, sim_matrix, tripadvisor_df):
    """
    Pour chaque place du train, on classe le test par similarité décroissante.
    On cherche la position de la 1ère place du test ayant le même typeR.
    Retourne la liste des erreurs (les requêtes sans match dans le test sont ignorées).
    """
    errors = []

    # Récupérer typeR pour toutes les places du test
    test_with_meta = test_df.merge(tripadvisor_df[['id', 'typeR']], left_on='idplace', right_on='id', how='left')
    train_with_meta = train_df.merge(tripadvisor_df[['id', 'typeR']], left_on='idplace', right_on='id', how='left')

    for i in range(len(train_df)):
        query_type = train_with_meta.iloc[i]['typeR']

        if pd.isna(query_type):
            continue  # pas de metadata pour cette requête

        # Vérifier qu'il existe au moins une place dans le test avec le même type
        test_types = test_with_meta['typeR'].values
        if query_type not in test_types:
            continue  # aucun match possible → on ignore

        # Classer le test par similarité décroissante
        scores = sim_matrix[i]  # vecteur de similarité avec le test
        ranked_indices = scores.argsort()[::-1]  # du plus similaire au moins similaire

        # Chercher la position du 1er match
        for rank, idx in enumerate(ranked_indices):
            if test_with_meta.iloc[idx]['typeR'] == query_type:
                errors.append(rank)  # rank=0 si premier résultat est bon
                break

    return errors


errors_l1 = ranking_error_level1(train_df, test_df, sim_train_test, tripadvisor)

mean_error_l1 = sum(errors_l1) / len(errors_l1)
print(f"Ranking Error Level 1 (TF-IDF)")
print(f"  Nb requêtes évaluées : {len(errors_l1)}")
print(f"  Mean Ranking Error   : {mean_error_l1:.2f}")
print(f"  Erreur min : {min(errors_l1)}, max : {max(errors_l1)}")
````

## Ranking Error Level 2 — basé sur `activiteSubCategorie` / `restaurantTypeCuisine`

Pour le Level 2 on utilise des attributs plus précis :
- **Attractions** → `activiteSubCategorie` (liste d'IDs)
- **Restaurants** → `restaurantTypeCuisine` (liste d'IDs)
- **Hotels** → `priceRange`

Un match est positif si **au moins un** ID en commun existe entre la requête et la place recommandée.

````python
def get_fine_category(row):
    """
    Retourne la liste de catégories fines selon le type de lieu.
    """
    type_r = row.get('typeR', None)
    if type_r == 'A' or type_r == 'AP':
        cats = row.get('activiteSubCategorie', [])
        return cats if isinstance(cats, list) else []
    elif type_r == 'R':
        cats = row.get('restaurantTypeCuisine', [])
        return cats if isinstance(cats, list) else []
    elif type_r == 'H':
        # Pour les hotels on utilise priceRange comme catégorie
        pr = row.get('priceRange', None)
        return [pr] if pd.notna(pr) else []
    return []


def categories_match(query_cats, candidate_cats):
    """Match si au moins une catégorie en commun."""
    if not query_cats or not candidate_cats:
        return False
    return len(set(query_cats) & set(candidate_cats)) > 0


def ranking_error_level2(train_df, test_df, sim_matrix, tripadvisor_df):
    errors = []

    cols = ['id', 'typeR', 'activiteSubCategorie', 'restaurantTypeCuisine', 'priceRange']
    # Garder seulement les colonnes disponibles
    available_cols = [c for c in cols if c in tripadvisor_df.columns]
    meta = tripadvisor_df[available_cols].copy()

    test_with_meta  = test_df.merge(meta, left_on='idplace', right_on='id', how='left')
    train_with_meta = train_df.merge(meta, left_on='idplace', right_on='id', how='left')

    for i in range(len(train_df)):
        query_row = train_with_meta.iloc[i]
        query_cats = get_fine_category(query_row)

        if not query_cats:
            continue  # pas de catégorie fine → on ignore

        # Vérifier qu'au moins une place du test partage une catégorie
        has_match_in_test = any(
            categories_match(query_cats, get_fine_category(test_with_meta.iloc[j]))
            for j in range(len(test_df))
        )
        if not has_match_in_test:
            continue

        # Classer le test par similarité
        scores = sim_matrix[i]
        ranked_indices = scores.argsort()[::-1]

        for rank, idx in enumerate(ranked_indices):
            candidate_cats = get_fine_category(test_with_meta.iloc[idx])
            if categories_match(query_cats, candidate_cats):
                errors.append(rank)
                break

    return errors


errors_l2 = ranking_error_level2(train_df, test_df, sim_train_test, tripadvisor)

mean_error_l2 = sum(errors_l2) / len(errors_l2)
print(f"Ranking Error Level 2 (TF-IDF)")
print(f"  Nb requêtes évaluées : {len(errors_l2)}")
print(f"  Mean Ranking Error   : {mean_error_l2:.2f}")
print(f"  Erreur min : {min(errors_l2)}, max : {max(errors_l2)}")
````

## Baseline BM25

BM25 est un modèle de recherche d'information classique, plus sophistiqué que TF-IDF car il normalise par la longueur des documents.  
C'est la baseline obligatoire du projet.

````python
# !pip install rank-bm25
from rank_bm25 import BM25Okapi

# BM25 attend une liste de listes de tokens
# On retokenise à partir de text_resume (split sur espaces)
train_corpus = [text.split() for text in train_df["text_resume"]]
test_corpus  = [text.split() for text in test_df["text_resume"]]

# On construit le modèle BM25 sur le test (c'est le corpus dans lequel on cherche)
bm25 = BM25Okapi(test_corpus)

print("BM25 construit sur le test corpus.")
````

````python
# Vérification rapide : recommandation BM25 pour une place du train
query_tokens = train_corpus[0]
bm25_scores  = bm25.get_scores(query_tokens)

best_idx = bm25_scores.argsort()[::-1][0]
print("Requête      :", get_place_name(train_df.iloc[0]['idplace']))
print("Recommandation BM25 :", get_place_name(test_df.iloc[best_idx]['idplace']))
````

### Ranking Error Level 1 — BM25

````python
def ranking_error_bm25_level1(train_df, test_df, bm25_model, train_corpus, tripadvisor_df):
    errors = []

    test_with_meta  = test_df.merge(tripadvisor_df[['id', 'typeR']], left_on='idplace', right_on='id', how='left')
    train_with_meta = train_df.merge(tripadvisor_df[['id', 'typeR']], left_on='idplace', right_on='id', how='left')

    for i in range(len(train_df)):
        query_type = train_with_meta.iloc[i]['typeR']
        if pd.isna(query_type):
            continue
        if query_type not in test_with_meta['typeR'].values:
            continue

        scores = bm25_model.get_scores(train_corpus[i])
        ranked_indices = scores.argsort()[::-1]

        for rank, idx in enumerate(ranked_indices):
            if test_with_meta.iloc[idx]['typeR'] == query_type:
                errors.append(rank)
                break

    return errors


errors_bm25_l1 = ranking_error_bm25_level1(train_df, test_df, bm25, train_corpus, tripadvisor)

mean_bm25_l1 = sum(errors_bm25_l1) / len(errors_bm25_l1)
print(f"Ranking Error Level 1 (BM25)")
print(f"  Nb requêtes évaluées : {len(errors_bm25_l1)}")
print(f"  Mean Ranking Error   : {mean_bm25_l1:.2f}")
````

### Ranking Error Level 2 — BM25

````python
def ranking_error_bm25_level2(train_df, test_df, bm25_model, train_corpus, tripadvisor_df):
    errors = []

    cols = ['id', 'typeR', 'activiteSubCategorie', 'restaurantTypeCuisine', 'priceRange']
    available_cols = [c for c in cols if c in tripadvisor_df.columns]
    meta = tripadvisor_df[available_cols].copy()

    test_with_meta  = test_df.merge(meta, left_on='idplace', right_on='id', how='left')
    train_with_meta = train_df.merge(meta, left_on='idplace', right_on='id', how='left')

    for i in range(len(train_df)):
        query_row  = train_with_meta.iloc[i]
        query_cats = get_fine_category(query_row)

        if not query_cats:
            continue

        has_match_in_test = any(
            categories_match(query_cats, get_fine_category(test_with_meta.iloc[j]))
            for j in range(len(test_df))
        )
        if not has_match_in_test:
            continue

        scores = bm25_model.get_scores(train_corpus[i])
        ranked_indices = scores.argsort()[::-1]

        for rank, idx in enumerate(ranked_indices):
            candidate_cats = get_fine_category(test_with_meta.iloc[idx])
            if categories_match(query_cats, candidate_cats):
                errors.append(rank)
                break

    return errors


errors_bm25_l2 = ranking_error_bm25_level2(train_df, test_df, bm25, train_corpus, tripadvisor)

mean_bm25_l2 = sum(errors_bm25_l2) / len(errors_bm25_l2)
print(f"Ranking Error Level 2 (BM25)")
print(f"  Nb requêtes évaluées : {len(errors_bm25_l2)}")
print(f"  Mean Ranking Error   : {mean_bm25_l2:.2f}")
````

## Comparaison finale : TF-IDF vs BM25

````python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame({
    'Modèle'            : ['TF-IDF + Cosine', 'BM25'],
    'Mean Error Level 1': [mean_error_l1, mean_bm25_l1],
    'Mean Error Level 2': [mean_error_l2, mean_bm25_l2],
})

print(results.to_string(index=False))

# Visualisation
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

for ax, level, col in zip(axes, ['Level 1', 'Level 2'], ['Mean Error Level 1', 'Mean Error Level 2']):
    ax.bar(results['Modèle'], results[col], color=['steelblue', 'coral'])
    ax.set_title(f'Mean Ranking Error — {level}')
    ax.set_ylabel('Mean Ranking Error (lower is better)')
    for i, v in enumerate(results[col]):
        ax.text(i, v + 0.3, f'{v:.1f}', ha='center', fontsize=11)

plt.tight_layout()
plt.savefig('comparison_tfidf_bm25.png', dpi=150)
plt.show()
print("\nRésultat : un Mean Ranking Error plus bas = meilleur modèle.")
````

## Interprétation des résultats

**Level 1 (typeR)** — évaluation grossière sur 4 types (H/R/A/AP) :  
Une erreur faible signifie que le modèle retrouve rapidement un lieu du même type général.

**Level 2 (sous-catégories)** — évaluation fine sur les catégories précises :  
L'erreur est naturellement plus élevée car le critère est plus strict. Un match est positif si au moins une sous-catégorie est commune.

**BM25 vs TF-IDF cosine** :  
BM25 normalise par la longueur des documents, ce qui peut avantager les lieux avec peu de reviews. TF-IDF cosine favorise les lieux avec beaucoup de reviews car leur vecteur est plus dense.  
Le modèle avec le Mean Ranking Error le plus bas est le meilleur.

Ces résultats confirment (ou infirment) l'hypothèse du projet : **des expériences similaires sont décrites avec des mots similaires**.
