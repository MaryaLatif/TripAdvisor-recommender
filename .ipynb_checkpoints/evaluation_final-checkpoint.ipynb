## Evaluation Setup

On recharge les colonnes nécessaires depuis le CSV original.

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

meta_full = pd.read_csv('data/Tripadvisor.csv', low_memory=False)

meta = meta_full[['id', 'typeR', 'activiteSubCategorie', 'restaurantType', 'priceRange']].copy()

# Convertir activiteSubCategorie en liste d'entiers
meta['activiteSubCategorie'] = meta['activiteSubCategorie'].apply(
    lambda s: [int(x.strip()) for x in str(s).split(',') if x.strip() not in ['nan', '']]
)

# Convertir restaurantType en liste d'entiers
meta['restaurantType'] = meta['restaurantType'].apply(
    lambda s: [int(x.strip()) for x in str(s).split(',') if x.strip() not in ['nan', '']]
)

print(meta.shape)
print(meta['typeR'].value_counts())
meta.head()
```

## Train / Test Split 50/50

On split les **places** (pas les reviews).  
- **Train** = les requêtes (ce qu'on cherche à recommander)  
- **Test** = le corpus dans lequel on recommande  

Les métadonnées servent **uniquement à l'évaluation**, jamais en entrée du modèle.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

train_df, test_df = train_test_split(reviews_for_tfidf, test_size=0.5, random_state=42)
train_df = train_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

print(f"Train : {len(train_df)} places")
print(f"Test  : {len(test_df)} places")

# TF-IDF : fit sur train uniquement, transform sur test
tfidf_eval   = TfidfVectorizer()
train_matrix = tfidf_eval.fit_transform(train_df['text_resume'])
test_matrix  = tfidf_eval.transform(test_df['text_resume'])

# Matrice de similarité : (nb_train x nb_test)
sim_matrix = cosine_similarity(train_matrix, test_matrix)
print(f"Matrice similarité : {sim_matrix.shape}")
```

## Ranking Error — Level 1 (`typeR`)

Pour chaque place du train, on classe tout le test par similarité décroissante.  
On note la **position** (0-indexée) de la première place du test avec le même `typeR`.  
- Erreur = 0 → recommandation parfaite  
- Erreur = k → il faut descendre à la position k pour trouver un bon match  
- Pas de match dans le test → requête ignorée

```python
train_meta = train_df.merge(meta[['id', 'typeR']], left_on='idplace', right_on='id', how='left')
test_meta  = test_df.merge(meta[['id', 'typeR']], left_on='idplace', right_on='id', how='left')

test_types = test_meta['typeR'].values  # tableau des types du test

errors_l1 = []

for i in range(len(train_df)):
    query_type = train_meta.iloc[i]['typeR']

    if pd.isna(query_type):
        continue
    if query_type not in test_types:
        continue

    ranked = sim_matrix[i].argsort()[::-1]

    for rank, idx in enumerate(ranked):
        if test_types[idx] == query_type:
            errors_l1.append(rank)
            break

mean_l1 = sum(errors_l1) / len(errors_l1)
median_l1 = sorted(errors_l1)[len(errors_l1) // 2]

print(f"Nb requêtes évaluées  : {len(errors_l1)}")
print(f"Mean Ranking Error L1 : {mean_l1:.2f}")
print(f"Median Ranking Error  : {median_l1}")
```

## Ranking Error — Level 2 (sous-catégories)

Évaluation plus fine selon le type de lieu :  
- **Attractions (A / AP)** → `activiteSubCategorie`  
- **Restaurants (R)** → `restaurantType`  
- **Hotels (H)** → `priceRange`  

**Match positif** = au moins une catégorie en commun entre la requête et la place recommandée.

```python
train_meta2 = train_df.merge(meta, left_on='idplace', right_on='id', how='left')
test_meta2  = test_df.merge(meta, left_on='idplace', right_on='id', how='left')


def get_cats(row):
    """Retourne les catégories fines selon le type de lieu."""
    t = row['typeR']
    if t in ('A', 'AP'):
        c = row['activiteSubCategorie']
        return c if isinstance(c, list) else []
    elif t == 'R':
        c = row['restaurantType']
        return c if isinstance(c, list) else []
    elif t == 'H':
        pr = row['priceRange']
        return [pr] if pd.notna(pr) else []
    return []


# Précalculer les catégories du test (évite de recalculer à chaque itération)
test_cats_list = [get_cats(test_meta2.iloc[j]) for j in range(len(test_df))]

errors_l2 = []

for i in range(len(train_df)):
    query_cats = get_cats(train_meta2.iloc[i])

    if not query_cats:
        continue

    query_set = set(query_cats)

    # Vérifier qu'un match existe dans le test
    if not any(query_set & set(c) for c in test_cats_list):
        continue

    ranked = sim_matrix[i].argsort()[::-1]

    for rank, idx in enumerate(ranked):
        if query_set & set(test_cats_list[idx]):
            errors_l2.append(rank)
            break

mean_l2 = sum(errors_l2) / len(errors_l2)
median_l2 = sorted(errors_l2)[len(errors_l2) // 2]

print(f"Nb requêtes évaluées  : {len(errors_l2)}")
print(f"Mean Ranking Error L2 : {mean_l2:.2f}")
print(f"Median Ranking Error  : {median_l2}")
```

## Visualisation

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, errors, title, color, mean_val in [
    (axes[0], errors_l1, 'Level 1 — typeR',           'steelblue', mean_l1),
    (axes[1], errors_l2, 'Level 2 — sous-catégories', 'coral',     mean_l2)
]:
    ax.hist(errors, bins=30, color=color, edgecolor='white')
    ax.axvline(mean_val, color='red', linestyle='--', label=f'Mean = {mean_val:.1f}')
    ax.set_title(f'Ranking Error Distribution — {title}')
    ax.set_xlabel('Position du 1er match (0 = parfait)')
    ax.set_ylabel('Nb requêtes')
    ax.legend()

plt.tight_layout()
plt.savefig('ranking_error_distribution.png', dpi=150)
plt.show()

print(f"\nRésumé :")
print(f"  Mean Ranking Error Level 1 : {mean_l1:.2f}")
print(f"  Mean Ranking Error Level 2 : {mean_l2:.2f}")
```

## Interprétation

**Level 1 (typeR)** — critère grossier sur 4 types (H/R/A/AP).  
Une erreur faible signifie que le vocabulaire des reviews est suffisamment distinctif entre types de lieux : un hôtel et un restaurant ne sont pas décrits avec les mêmes mots.

**Level 2 (sous-catégories)** — critère fin.  
L'erreur est naturellement plus élevée car le critère est plus strict. Un match est positif si au moins une sous-catégorie est commune.

Ces résultats permettent de valider l'hypothèse centrale du projet : **des expériences similaires sont décrites avec des mots similaires**.
